## Objetivo
Implementar, comparar y analizar cuatro modelos fundamentales de clasificación supervisada — **Regresión Logística, Árbol de Decisión, SVM y KNN** — sobre el dataset **Wine Quality** de UCI, evaluando sus fortalezas, limitaciones y sensibilidad al preprocesamiento.

## 1. Configuración del entorno

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection import (
    train_test_split, cross_val_score, StratifiedKFold, GridSearchCV
)
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    classification_report, confusion_matrix,
    ConfusionMatrixDisplay, RocCurveDisplay,
    accuracy_score, f1_score
)
from sklearn.inspection import DecisionBoundaryDisplay

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['figure.dpi'] = 100

RANDOM_STATE = 42
print('Entorno configurado correctamente.')
```

---
## 2. Carga y comprensión del dataset

Utilizamos el **Wine Quality (Red)** del repositorio UCI Machine Learning. Contiene 1599 muestras de vino tinto portugués con 11 atributos fisicoquímicos y una puntuación de calidad (3–8).

Para construir un problema de **clasificación binaria**, definimos:
- **Clase 1 (Bueno):** calidad ≥ 7
- **Clase 0 (Estándar):** calidad < 7

Esta binarización genera un **problema desbalanceado**, lo que añade un reto realista al modelado.

```python
# Carga directa desde UCI
URL = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv'

df = pd.read_csv(URL, sep=';')
print(f'Dataset cargado: {df.shape[0]} registros, {df.shape[1]} columnas')
df.head()
```

```python
# Distribución original de la variable 'quality'
print('Distribución de calidad original:')
print(df['quality'].value_counts().sort_index())

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

# Histograma original
sns.countplot(data=df, x='quality', ax=axes[0], palette='viridis')
axes[0].set_title('Distribución original de calidad')
axes[0].set_xlabel('Puntuación de calidad')

# Binarización
df['quality_label'] = (df['quality'] >= 7).astype(int)

counts = df['quality_label'].value_counts()
axes[1].bar(['Estándar (0)', 'Bueno (1)'], counts.values,
            color=['#3498db', '#e74c3c'])
for i, v in enumerate(counts.values):
    axes[1].text(i, v + 10, f'{v} ({v/len(df)*100:.1f}%)', ha='center', fontweight='bold')
axes[1].set_title('Clasificación binaria (umbral: quality ≥ 7)')
axes[1].set_ylabel('Cantidad')

plt.tight_layout()
plt.show()

print(f'\nRatio de desbalance: {counts[0]/counts[1]:.1f}:1 (Estándar vs Bueno)')
```

---
## 3. Análisis exploratorio orientado a clasificación

```python
# 3.1  Estadísticas descriptivas por clase
features = df.columns.drop(['quality', 'quality_label'])

comparison = df.groupby('quality_label')[features].mean().T
comparison.columns = ['Estándar (0)', 'Bueno (1)']
comparison['Diferencia %'] = ((comparison['Bueno (1)'] - comparison['Estándar (0)']) / comparison['Estándar (0)'] * 100).round(1)
comparison.style.background_gradient(subset=['Diferencia %'], cmap='RdYlGn')
```

```python
# 3.2  Distribución de features por clase (boxplots)
fig, axes = plt.subplots(3, 4, figsize=(18, 12))
axes = axes.flatten()

for i, col in enumerate(features):
    sns.boxplot(data=df, x='quality_label', y=col, ax=axes[i],
                palette=['#3498db', '#e74c3c'], hue='quality_label', legend=False)
    axes[i].set_title(col, fontsize=10)
    axes[i].set_xlabel('')
    axes[i].set_xticklabels(['Estándar', 'Bueno'])

# Ocultar subplot vacío
axes[-1].set_visible(False)
fig.suptitle('Distribución de features por clase', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

```python
# 3.3  Correlación entre features
plt.figure(figsize=(10, 8))
corr = df[features].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, linewidths=0.5, annot_kws={'size': 8})
plt.title('Matriz de correlación — Features')
plt.tight_layout()
plt.show()
```

---
## 4. Preparación de datos

```python
# 4.1  Separación features / target
X = df[features].copy()
y = df['quality_label'].copy()

# 4.2  Split estratificado 80/20
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

print(f'Train: {X_train.shape[0]} muestras  |  Test: {X_test.shape[0]} muestras')
print(f'Proporción clase 1 en train: {y_train.mean():.3f}')
print(f'Proporción clase 1 en test:  {y_test.mean():.3f}')
```

```python
# 4.3  Verificación de escalas (justifica la estandarización)
print('Rangos de las features (train):')
print('='*50)
for col in features:
    mn, mx = X_train[col].min(), X_train[col].max()
    print(f'{col:30s}  [{mn:8.2f}, {mx:8.2f}]  rango: {mx-mn:.2f}')

print('\n→ Las escalas difieren significativamente entre features.')
print('  Esto afecta directamente a SVM y KNN (basados en distancias).')
print('  Logistic Regression también se beneficia de la estandarización.')
```

---
## 5. Modelo 1 — Regresión Logística

**Fundamento:** Modelo lineal que estima la probabilidad de pertenencia a una clase mediante la función sigmoide. Ideal como **baseline** por su interpretabilidad y eficiencia.

```python
# 5.1  Pipeline con estandarización
pipe_lr = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', LogisticRegression(max_iter=1000, random_state=RANDOM_STATE,
                               class_weight='balanced'))  # Compensar desbalance
])

# Validación cruzada
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
scores_lr = cross_val_score(pipe_lr, X_train, y_train, cv=cv, scoring='f1')
print(f'Logistic Regression — F1 CV: {scores_lr.mean():.4f} ± {scores_lr.std():.4f}')

# Entrenamiento final
pipe_lr.fit(X_train, y_train)
y_pred_lr = pipe_lr.predict(X_test)
```

```python
# 5.2  Coeficientes del modelo (interpretabilidad)
coefs = pd.Series(
    pipe_lr.named_steps['clf'].coef_[0],
    index=features
).sort_values()

fig, ax = plt.subplots(figsize=(8, 5))
colors = ['#e74c3c' if c < 0 else '#2ecc71' for c in coefs.values]
coefs.plot(kind='barh', ax=ax, color=colors)
ax.set_title('Regresión Logística — Coeficientes estandarizados')
ax.set_xlabel('Coeficiente (impacto en la predicción)')
ax.axvline(x=0, color='black', linewidth=0.8)
plt.tight_layout()
plt.show()

print('\nInterpretación: Un coeficiente positivo alto indica que valores mayores')
print('de esa feature incrementan la probabilidad de clasificar como "Bueno".')
```

---
## 6. Modelo 2 — Árbol de Decisión

**Fundamento:** Modelo no lineal que particiona el espacio de features mediante reglas if-else jerárquicas. Altamente interpretable, pero propenso al sobreajuste si no se controla la profundidad.

```python
# 6.1  Pipeline (no requiere estandarización)
pipe_dt = Pipeline([
    ('clf', DecisionTreeClassifier(
        max_depth=5, min_samples_leaf=10,
        class_weight='balanced', random_state=RANDOM_STATE
    ))
])

scores_dt = cross_val_score(pipe_dt, X_train, y_train, cv=cv, scoring='f1')
print(f'Decision Tree — F1 CV: {scores_dt.mean():.4f} ± {scores_dt.std():.4f}')

pipe_dt.fit(X_train, y_train)
y_pred_dt = pipe_dt.predict(X_test)
```

```python
# 6.2  Visualización del árbol
fig, ax = plt.subplots(figsize=(20, 8))
plot_tree(
    pipe_dt.named_steps['clf'],
    feature_names=list(features),
    class_names=['Estándar', 'Bueno'],
    filled=True, rounded=True, fontsize=8, ax=ax,
    max_depth=3  # Mostrar solo 3 niveles para legibilidad
)
ax.set_title('Árbol de Decisión (primeros 3 niveles)', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# 6.3  Importancia de features (Gini importance)
importances = pd.Series(
    pipe_dt.named_steps['clf'].feature_importances_,
    index=features
).sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(8, 5))
importances.plot(kind='barh', ax=ax, color='#3498db')
ax.set_title('Árbol de Decisión — Importancia de features (Gini)')
ax.set_xlabel('Importancia relativa')
plt.tight_layout()
plt.show()
```

```python
# 6.4  Efecto de la profundidad en sobreajuste/subajuste
depths = range(1, 21)
train_scores = []
val_scores = []

for d in depths:
    dt = DecisionTreeClassifier(max_depth=d, class_weight='balanced',
                                 random_state=RANDOM_STATE)
    dt.fit(X_train, y_train)
    train_scores.append(f1_score(y_train, dt.predict(X_train)))
    cv_s = cross_val_score(dt, X_train, y_train, cv=cv, scoring='f1')
    val_scores.append(cv_s.mean())

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(depths, train_scores, 'o-', label='Train F1', color='#2ecc71')
ax.plot(depths, val_scores, 's-', label='Validación F1 (CV)', color='#e74c3c')
ax.axvline(x=5, color='gray', linestyle='--', alpha=0.7, label='max_depth=5 (seleccionado)')
ax.set_xlabel('Profundidad máxima')
ax.set_ylabel('F1-Score')
ax.set_title('Árbol de Decisión — Sobreajuste vs Profundidad')
ax.legend()
ax.set_xticks(depths)
plt.tight_layout()
plt.show()

print('Observación: cuando la brecha entre train y validación crece,')
print('el modelo está sobreajustando (memorizando el train set).')
```

---
## 7. Modelo 3 — Support Vector Machine (SVM)

**Fundamento:** Busca el hiperplano que maximiza el margen de separación entre clases. Con funciones kernel puede manejar fronteras no lineales. Es **sensible a la escala** de las features.

```python
# 7.1  Comparación de kernels
kernels = ['linear', 'rbf', 'poly']
svm_results = {}

for kernel in kernels:
    pipe_svm = Pipeline([
        ('scaler', StandardScaler()),
        ('clf', SVC(kernel=kernel, class_weight='balanced',
                     random_state=RANDOM_STATE, probability=True))
    ])
    scores = cross_val_score(pipe_svm, X_train, y_train, cv=cv, scoring='f1')
    svm_results[kernel] = scores
    print(f'SVM ({kernel:6s}) — F1 CV: {scores.mean():.4f} ± {scores.std():.4f}')

# Seleccionar el mejor kernel
best_kernel = max(svm_results, key=lambda k: svm_results[k].mean())
print(f'\nMejor kernel: {best_kernel}')
```

```python
# 7.2  Entrenamiento con el mejor kernel
pipe_svm = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', SVC(kernel=best_kernel, class_weight='balanced',
                 random_state=RANDOM_STATE, probability=True))
])

pipe_svm.fit(X_train, y_train)
y_pred_svm = pipe_svm.predict(X_test)
scores_svm = svm_results[best_kernel]
```

```python
# 7.3  Impacto de la estandarización en SVM
# SVM SIN estandarizar
svm_no_scale = SVC(kernel=best_kernel, class_weight='balanced',
                    random_state=RANDOM_STATE)
scores_no_scale = cross_val_score(svm_no_scale, X_train, y_train, cv=cv, scoring='f1')

print('Impacto de la estandarización en SVM:')
print(f'  CON StandardScaler: F1 = {scores_svm.mean():.4f}')
print(f'  SIN StandardScaler: F1 = {scores_no_scale.mean():.4f}')
print(f'  Diferencia: {(scores_svm.mean() - scores_no_scale.mean())*100:+.2f} puntos porcentuales')
print('\n→ La estandarización es CRÍTICA para modelos basados en distancias.')
```

---
## 8. Modelo 4 — K-Nearest Neighbors (KNN)

**Fundamento:** Clasifica según la mayoría de votos de los K vecinos más cercanos. No paramétrico, sensible a la escala y a la dimensionalidad. El valor de K controla el trade-off bias-varianza.

```python
# 8.1  Búsqueda del K óptimo
k_range = range(1, 31)
k_scores = []

for k in k_range:
    pipe_knn = Pipeline([
        ('scaler', StandardScaler()),
        ('clf', KNeighborsClassifier(n_neighbors=k))
    ])
    scores = cross_val_score(pipe_knn, X_train, y_train, cv=cv, scoring='f1')
    k_scores.append(scores.mean())

best_k = k_range[np.argmax(k_scores)]

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(k_range, k_scores, 'o-', markersize=4, color='#8e44ad')
ax.axvline(x=best_k, color='red', linestyle='--', alpha=0.7,
           label=f'Mejor K = {best_k} (F1 = {max(k_scores):.4f})')
ax.set_xlabel('Número de vecinos (K)')
ax.set_ylabel('F1-Score (CV)')
ax.set_title('KNN — Búsqueda del K óptimo')
ax.legend()
ax.set_xticks(range(0, 31, 2))
plt.tight_layout()
plt.show()

print(f'K óptimo: {best_k}')
```

```python
# 8.2  Entrenamiento con K óptimo
pipe_knn = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', KNeighborsClassifier(n_neighbors=best_k))
])

scores_knn = cross_val_score(pipe_knn, X_train, y_train, cv=cv, scoring='f1')
print(f'KNN (K={best_k}) — F1 CV: {scores_knn.mean():.4f} ± {scores_knn.std():.4f}')

pipe_knn.fit(X_train, y_train)
y_pred_knn = pipe_knn.predict(X_test)
```

## 9. Análisis comparativo de modelos

```python
# 9.1  Tabla resumen
models = {
    'Logistic Regression': (pipe_lr, y_pred_lr, scores_lr),
    'Decision Tree':       (pipe_dt, y_pred_dt, scores_dt),
    f'SVM ({best_kernel})': (pipe_svm, y_pred_svm, scores_svm),
    f'KNN (K={best_k})':   (pipe_knn, y_pred_knn, scores_knn)
}

summary = []
for name, (pipe, y_pred, cv_scores) in models.items():
    summary.append({
        'Modelo': name,
        'F1 CV (mean)': f'{cv_scores.mean():.4f}',
        'F1 CV (std)': f'{cv_scores.std():.4f}',
        'Accuracy Test': f'{accuracy_score(y_test, y_pred):.4f}',
        'F1 Test': f'{f1_score(y_test, y_pred):.4f}'
    })

summary_df = pd.DataFrame(summary)
print('='*80)
print('COMPARACIÓN DE MODELOS')
print('='*80)
summary_df
```

```python
# 9.2  Boxplot comparativo de validación cruzada
fig, ax = plt.subplots(figsize=(10, 5))
cv_data = [scores_lr, scores_dt, scores_svm, scores_knn]
labels = list(models.keys())
bp = ax.boxplot(cv_data, tick_labels=labels, patch_artist=True)
colors_box = ['#3498db', '#2ecc71', '#e74c3c', '#8e44ad']
for patch, color in zip(bp['boxes'], colors_box):
    patch.set_facecolor(color)
    patch.set_alpha(0.6)
ax.set_ylabel('F1-Score')
ax.set_title('Comparación de modelos — Validación cruzada (5-fold)')
plt.tight_layout()
plt.show()
```

```python
# 9.3  Matrices de confusión comparativas
fig, axes = plt.subplots(1, 4, figsize=(20, 4))

for ax, (name, (pipe, y_pred, _)) in zip(axes, models.items()):
    ConfusionMatrixDisplay.from_predictions(
        y_test, y_pred,
        display_labels=['Estándar', 'Bueno'],
        cmap='Blues', ax=ax
    )
    ax.set_title(name, fontsize=10)

fig.suptitle('Matrices de confusión — Conjunto de test', fontsize=13, y=1.03)
plt.tight_layout()
plt.show()
```

```python
# 9.4  Curvas ROC comparativas
fig, ax = plt.subplots(figsize=(7, 6))

for name, (pipe, _, _) in models.items():
    RocCurveDisplay.from_estimator(pipe, X_test, y_test, ax=ax, name=name)

ax.plot([0, 1], [0, 1], 'k--', alpha=0.5, label='Aleatorio')
ax.set_title('Curvas ROC — Comparación de modelos')
ax.legend(loc='lower right', fontsize=9)
plt.tight_layout()
plt.show()
```

```python
# 9.5  Classification report detallado del mejor modelo
best_model_name = max(models, key=lambda k: float(models[k][2].mean()))
best_pipe, best_pred, best_cv = models[best_model_name]

print(f'\nMejor modelo: {best_model_name}')
print('='*60)
print(classification_report(y_test, best_pred,
                             target_names=['Estándar', 'Bueno']))
```

| Modelo | Fortalezas | Limitaciones | ¿Cuándo usarlo? |
|---|---|---|---|
| **Regresión Logística** | Rápido, interpretable (coeficientes), probabilístico | Asume relación lineal entre features y log-odds | Baseline; cuando la interpretabilidad es prioritaria |

# 11. Ejercicios propuestos
# Ejercicio 1 — GridSearchCV para SVM

**Objetivo:** buscar los mejores hiperparámetros de un SVM con `GridSearchCV` y compararlo contra el SVM base de la sesión.

**Paso 1 — Definir el pipeline y la grilla de hiperparámetros**

Usamos un `Pipeline` con `StandardScaler` + `SVC`, para que el escalado se ajuste solo con los datos de entrenamiento de cada fold (evita fuga de datos). En la grilla, los parámetros del clasificador llevan el prefijo `clf__`.

La grilla tiene 4 (C) × 4 (gamma) × 2 (kernel) = **32 combinaciones**, y con CV de 5 folds son **160 entrenamientos**. Se desactiva `probability=True` porque no se necesita para la búsqueda y hace el proceso mucho más lento.

```python
pipe_svm_grid = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', SVC(class_weight='balanced', random_state=RANDOM_STATE))
])

param_grid = {
    'clf__C': [0.1, 1, 10, 100],
    'clf__gamma': ['scale', 'auto', 0.01, 0.1],
    'clf__kernel': ['rbf', 'poly']
}

n_comb = len(param_grid['clf__C']) * len(param_grid['clf__gamma']) * len(param_grid['clf__kernel'])
print(f'Combinaciones a evaluar: {n_comb}')
print(f'Entrenamientos totales (5 folds): {n_comb * 5}')
```

**Paso 2 — Ejecutar GridSearchCV**

Se usa la métrica `f1` (clase positiva = "Bueno"), el mismo `cv` estratificado de la sesión y `n_jobs=-1` para paralelizar. Con `refit=True` (por defecto) el mejor modelo se reentrena con todo el train.

```python
grid_svm = GridSearchCV(
    estimator=pipe_svm_grid,
    param_grid=param_grid,
    scoring='f1',
    cv=cv,
    n_jobs=-1,
    verbose=1,
    return_train_score=True
)

grid_svm.fit(X_train, y_train)
print('Búsqueda finalizada.')
```

**Paso 3 — Reportar los mejores hiperparámetros y el F1 de validación cruzada**

```python
print('='*60)
print('RESULTADOS DE GridSearchCV')
print('='*60)
print('Mejores hiperparámetros:')
for param, valor in grid_svm.best_params_.items():
    print(f'  {param.replace("clf__", ""):8s} = {valor}')

print(f'\nMejor F1 (CV, 5-fold): {grid_svm.best_score_:.4f}')

# Top 10 combinaciones
results = pd.DataFrame(grid_svm.cv_results_)
top10 = (results[['param_clf__kernel', 'param_clf__C', 'param_clf__gamma',
                  'mean_train_score', 'mean_test_score', 'std_test_score', 'rank_test_score']]
         .sort_values('rank_test_score')
         .head(10)
         .rename(columns={
             'param_clf__kernel': 'kernel',
             'param_clf__C': 'C',
             'param_clf__gamma': 'gamma',
             'mean_train_score': 'F1 train',
             'mean_test_score': 'F1 CV',
             'std_test_score': 'std CV',
             'rank_test_score': 'ranking'
         }))
top10
```

**Paso 4 — Visualizar el efecto de los hiperparámetros**

Un mapa de calor del F1 de validación por `C` y `gamma`, separado por kernel, permite ver qué zonas de la grilla funcionan mejor.

```python
results['kernel'] = results['param_clf__kernel']
results['C'] = results['param_clf__C']
results['gamma'] = results['param_clf__gamma'].astype(str)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
for ax, kernel in zip(axes, ['rbf', 'poly']):
    pivot = (results[results['kernel'] == kernel]
             .pivot(index='C', columns='gamma', values='mean_test_score'))
    sns.heatmap(pivot, annot=True, fmt='.3f', cmap='YlGnBu', ax=ax)
    ax.set_title(f'F1 CV — kernel = {kernel}')
    ax.set_xlabel('gamma')
    ax.set_ylabel('C')

plt.tight_layout()
plt.show()
```

**Paso 5 — Evaluar el mejor modelo en test y compararlo con el SVM base**

Se compara el SVM optimizado (`grid_svm.best_estimator_`) contra el SVM base de la sesión (`pipe_svm`, kernel `poly`, con hiperparámetros por defecto). Se contrastan tanto el F1 de validación cruzada como las métricas en el conjunto de test.

# Ejercicio 2 — Clasificación multiclase

**Objetivo:** en lugar de binarizar `quality`, usar las clases originales (3–8), entrenar los 4 modelos y analizar el cambio de rendimiento, las clases más difíciles y la métrica F1 más adecuada.

**Paso 1 — Definir el target multiclase y hacer el split estratificado**

Se usa `quality` directamente como variable objetivo. Las clases 3 (10 muestras) y 8 (18 muestras) son muy escasas. El split estratificado mantiene su proporción en train y test, y aún deja al menos 5 muestras por clase en train, lo que permite usar `StratifiedKFold` con 5 folds.

```python
y_multi = df['quality'].copy()
clases = sorted(y_multi.unique())

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
    X, y_multi, test_size=0.20, random_state=RANDOM_STATE, stratify=y_multi
)

dist = pd.DataFrame({
    'Total': y_multi.value_counts().sort_index(),
    'Train': y_train_m.value_counts().sort_index(),
    'Test': y_test_m.value_counts().sort_index()
})
dist['% del total'] = (dist['Total'] / len(y_multi) * 100).round(1)
print(f'Clases: {clases}')
display(dist)
```

**Paso 2 — Buscar el K óptimo de KNN para el problema multiclase**

El K=1 de la sesión se eligió para el problema binario, así que se vuelve a buscar. Como métrica se usa `f1_macro`, que promedia el F1 de cada clase por igual.

```python
k_range_m = range(1, 31)
k_scores_m = []

for k in k_range_m:
    pipe_k = Pipeline([
        ('scaler', StandardScaler()),
        ('clf', KNeighborsClassifier(n_neighbors=k))
    ])
    s = cross_val_score(pipe_k, X_train_m, y_train_m, cv=cv, scoring='f1_macro')
    k_scores_m.append(s.mean())

best_k_m = k_range_m[np.argmax(k_scores_m)]

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(k_range_m, k_scores_m, 'o-', markersize=4, color='#8e44ad')
ax.axvline(x=best_k_m, color='red', linestyle='--', alpha=0.7,
           label=f'Mejor K = {best_k_m} (F1 macro = {max(k_scores_m):.4f})')
ax.set_xlabel('Número de vecinos (K)')
ax.set_ylabel('F1 macro (CV)')
ax.set_title('KNN multiclase — Búsqueda del K óptimo')
ax.legend()
plt.tight_layout()
plt.show()

print(f'K óptimo (multiclase): {best_k_m}')
```

**Paso 3 — Definir los 4 modelos para multiclase**

Se mantienen las mismas configuraciones de la sesión (con `class_weight='balanced'` donde aplica). Para el SVM se usa el kernel elegido en la sesión (`best_kernel`) y se omite `probability=True` porque no se necesita. La regresión logística de scikit-learn maneja multiclase de forma nativa (softmax).

```python
modelos_m = {
    'Logistic Regression': Pipeline([
        ('scaler', StandardScaler()),
        ('clf', LogisticRegression(max_iter=2000, class_weight='balanced',
                                   random_state=RANDOM_STATE))
    ]),
    'Decision Tree': Pipeline([
        ('clf', DecisionTreeClassifier(max_depth=5, min_samples_leaf=10,
                                       class_weight='balanced', random_state=RANDOM_STATE))
    ]),
    f'SVM ({best_kernel})': Pipeline([
        ('scaler', StandardScaler()),
        ('clf', SVC(kernel=best_kernel, class_weight='balanced', random_state=RANDOM_STATE))
    ]),
    f'KNN (K={best_k_m})': Pipeline([
        ('scaler', StandardScaler()),
        ('clf', KNeighborsClassifier(n_neighbors=best_k_m))
    ])
}
print('Modelos definidos:', list(modelos_m.keys()))
```

**Paso 4 — Validación cruzada y predicciones en test**

Se calcula el F1 macro por validación cruzada (5-fold estratificado) y luego se entrena cada modelo con todo el train para predecir en test.

```python
cv_m = {}
pred_m = {}

for nombre, modelo in modelos_m.items():
    cv_m[nombre] = cross_val_score(modelo, X_train_m, y_train_m, cv=cv, scoring='f1_macro')
    modelo.fit(X_train_m, y_train_m)
    pred_m[nombre] = modelo.predict(X_test_m)
    print(f'{nombre:22s} — F1 macro CV: {cv_m[nombre].mean():.4f} ± {cv_m[nombre].std():.4f}')
```

**Paso 5 — Tabla comparativa con las tres variantes de F1 (macro, micro, weighted)**

- **Macro:** promedio simple del F1 de cada clase (todas pesan igual).
- **Micro:** calcula TP, FP y FN globales. En clasificación multiclase de una sola etiqueta, **es igual al accuracy**.
- **Weighted:** promedio de los F1 por clase ponderado por el soporte (número de muestras) de cada una.

```python
filas = []
for nombre, y_p in pred_m.items():
    filas.append({
        'Modelo': nombre,
        'Accuracy': accuracy_score(y_test_m, y_p),
        'F1 macro': f1_score(y_test_m, y_p, average='macro', zero_division=0),
        'F1 micro': f1_score(y_test_m, y_p, average='micro', zero_division=0),
        'F1 weighted': f1_score(y_test_m, y_p, average='weighted', zero_division=0),
        'F1 macro CV': cv_m[nombre].mean()
    })

tabla_m = pd.DataFrame(filas).set_index('Modelo').round(4)
print('='*80)
print('COMPARACIÓN MULTICLASE (test)')
print('='*80)
display(tabla_m.style.background_gradient(cmap='YlGn', axis=0))
```

**Paso 6 — ¿Cómo cambia el rendimiento respecto al caso binario?**

Se pone lado a lado el F1 de la sesión (binario, clase "Bueno") y el F1 macro multiclase. **Ojo:** no son estrictamente comparables, porque el F1 binario mide una sola clase positiva y el F1 macro promedia 6 clases (incluyendo las más raras). Aun así, muestra el aumento de dificultad.

```python
binario = {
    'Logistic Regression': (accuracy_score(y_test, y_pred_lr), f1_score(y_test, y_pred_lr)),
    'Decision Tree':       (accuracy_score(y_test, y_pred_dt), f1_score(y_test, y_pred_dt)),
    f'SVM ({best_kernel})': (accuracy_score(y_test, y_pred_svm), f1_score(y_test, y_pred_svm)),
    f'KNN (K={best_k})':   (accuracy_score(y_test, y_pred_knn), f1_score(y_test, y_pred_knn)),
}

nombres = list(pred_m.keys())
comp = pd.DataFrame({
    'Modelo': nombres,
    'Accuracy binario': [v[0] for v in binario.values()],
    'F1 binario (Bueno)': [v[1] for v in binario.values()],
    'Accuracy multiclase': [accuracy_score(y_test_m, pred_m[n]) for n in nombres],
    'F1 macro multiclase': [f1_score(y_test_m, pred_m[n], average='macro', zero_division=0) for n in nombres],
}).set_index('Modelo').round(4)
comp['Δ Accuracy'] = (comp['Accuracy multiclase'] - comp['Accuracy binario']).round(4)
display(comp)

# Gráfico comparativo
x = np.arange(len(nombres))
w = 0.2
fig, ax = plt.subplots(figsize=(12, 5))
ax.bar(x - 1.5*w, comp['Accuracy binario'], w, label='Accuracy binario', color='#3498db')
ax.bar(x - 0.5*w, comp['Accuracy multiclase'], w, label='Accuracy multiclase', color='#2980b9')
ax.bar(x + 0.5*w, comp['F1 binario (Bueno)'], w, label='F1 binario (Bueno)', color='#e74c3c')
ax.bar(x + 1.5*w, comp['F1 macro multiclase'], w, label='F1 macro multiclase', color='#c0392b')
ax.set_xticks(x)
ax.set_xticklabels(nombres, rotation=10)
ax.set_ylabel('Puntuación')
ax.set_title('Binario vs. multiclase')
ax.legend()
plt.tight_layout()
plt.show()
```

**Paso 7 — Matrices de confusión multiclase**

Se grafican con `labels` fijos (3–8) para que todas las clases aparezcan aunque un modelo nunca las prediga. Las celdas cercanas a la diagonal indican errores entre clases contiguas.

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 12))
axes = axes.flatten()

for ax, (nombre, y_p) in zip(axes, pred_m.items()):
    ConfusionMatrixDisplay.from_predictions(
        y_test_m, y_p, labels=clases, cmap='Blues', ax=ax, colorbar=False
    )
    ax.set_title(nombre)

fig.suptitle('Matrices de confusión multiclase — Test', fontsize=14, y=1.0)
plt.tight_layout()
plt.show()
```

**Paso 8 — ¿Qué clases son más difíciles de distinguir?**

Se calcula el F1 **por clase** para cada modelo y se muestra en un mapa de calor. Las clases con F1 bajo (cercano a 0) son las más difíciles. Se incluye el soporte de cada clase en test para contextualizar.

```python
f1_por_clase = pd.DataFrame(
    {n: f1_score(y_test_m, y_p, labels=clases, average=None, zero_division=0)
     for n, y_p in pred_m.items()},
    index=[f'Calidad {c}' for c in clases]
)
f1_por_clase['Soporte (test)'] = [int((y_test_m == c).sum()) for c in clases]

plt.figure(figsize=(10, 5))
sns.heatmap(f1_por_clase.drop(columns='Soporte (test)'), annot=True, fmt='.2f',
            cmap='RdYlGn', vmin=0, vmax=1)
plt.title('F1 por clase y por modelo (test)')
plt.tight_layout()
plt.show()

display(f1_por_clase.round(3))
```

**Paso 9 — Reporte de clasificación del mejor modelo (por F1 macro en test)**

```python
mejor_m = tabla_m['F1 macro'].idxmax()
print(f'Mejor modelo multiclase (F1 macro en test): {mejor_m}')
print('='*70)
print(classification_report(y_test_m, pred_m[mejor_m], labels=clases,
                            target_names=[f'Calidad {c}' for c in clases],
                            zero_division=0))
```

### Paso 10 — Análisis y justificación de la métrica F1

**Rendimiento respecto al caso binario:** completar con lo observado (el accuracy y el F1 suelen caer porque se pasa de 2 a 6 clases, con fronteras difusas entre puntuaciones vecinas).

**Clases más difíciles:** completar con el mapa de calor. Se espera que sean las extremas (3, 4 y 8) por tener muy pocas muestras, y que las clases centrales (5 y 6) se confundan entre sí porque las diferencias fisicoquímicas entre vinos de calidad contigua son sutiles.

**¿Qué F1 es más adecuado?**

| Variante | Comportamiento | Adecuada aquí |
|---|---|---|
| **Micro** | Equivale al accuracy: domina la información de las clases 5 y 6 y oculta el desempeño en las raras | No |
| **Weighted** | Pondera por soporte, así que también está dominado por 5 y 6 | Parcialmente (refleja el desempeño "típico") |
| **Macro** | Trata todas las clases por igual: penaliza a los modelos que ignoran las clases minoritarias | **Sí** |

**Justificación:** el dataset está fuertemente desbalanceado (clases 5 y 6 concentran ~82% de los datos) y las clases minoritarias (3, 4, 7, 8) suelen ser las de mayor interés. Por eso el **F1 macro** es la métrica más adecuada: un modelo que solo acierta 5 y 6 obtendría un micro/weighted alto pero un macro bajo, lo que evidencia su falla. Se recomienda reportar macro como métrica principal y weighted como complemento.
